In this markdown I put some root command to create histogram and so on with random numbers.

#Create the random variables between 0,1. if you don't specify the seed, it will be the same every time you restart the program.

$\text{gRandom->Uniform()}$

#Create an histogram, the argument in input are ("name", "title", n of bin, start, stop)

$\text{TH1F isto("name", "title", 400, -2, 2)}$

#populate the histogram with the random variables

$\text{for(int i=0; i<100000; i++) isto.Fill(gRandom->Uniform())}$

#Draw the histogram

$\text{isto.Draw()}$

#Populate another histogram with a power of the uniform distribution

$\text{TH1F isto2("name", "title", 400, -2, 2)}$
$\text{for(int i=0; i<100000; i++) isto2.Fill(pow(gRandom->Uniform(), 1./3))}$
$\text{isto2.Draw()}$
