```python
import numpy as np
tac=10
#ricorda di mettere in ordine i tuoi dati: quanto il primo rivelatore fa da sè va per primo ecc.
pmt=np.array([
    [30928    ,   9357  ,      67824,       384   ,          206],
    [29944   ,    8812    ,    68015   ,    325    ,         176],
    [30930   ,    9078    ,    67241    ,   362   ,          199],#1-2 fanno doppie
    [30618   ,    9332   ,     65940  ,     374     ,        181],

    [31008    ,   9114    ,    64986  ,     243   ,          162],
    [31150    ,   9131  ,      64319 ,      267   ,          181],
    [30143     ,  9128   ,     60955  ,     269    ,         188],#0-2 fanno doppie
    [31085   ,    9218   ,     67896     ,  265       ,      191],

    [30935     ,  9294   ,     65830     ,  307       ,      187], 
    [31272  ,     9066   ,     67552   ,    296          ,   185],#0-1 fanno doppie
    [30685     ,  9205   ,     65568    ,   313          ,   188],
    [30667   ,    8993 ,       66876 ,      296 ,            173]
])#12*5 array: 4 eventi, 3 combinazioni e #1, #2, #3, #doppie12, #triple


def efficienza(p):
    tac=10 #tempo di acquisizione in s
    tdiscr=40e-9
    tsovr=2e-9
    nev=4 #numero di eventi per combinazione
    comb=3 #numero di combinazione
    eff=[]#efficienza del terzo, secondo e primo rivelatore
    
    f=p[:,0:3]/tac



    for i in range(comb):
        db1=f[i*nev:(i+1)*nev, (i+1)%comb]*f[i*nev:(i+1)*nev, (i+2)%comb]*(tdiscr*2-tsovr)*tac
        eff1=p[i*nev:(i+1)*nev,4]/(p[i*nev:(i+1)*nev,3]-db1)
        var=eff1*(1-eff1)/p[i*nev:(i+1)*nev,3]
        media=np.mean(eff1)
        eff.append(media)
        std=np.sqrt(var.sum()/nev)
        eff.append(std)
        print(f"le doppie accidentali per il rivelatore {(i+1)%3} e {(i+2)%3} sono {db1}")

    return f"L'efficienza del rivelatore 0 è {eff[0]:.2f} {eff[1]:.2f}, 1 {eff[2]:.2f} {eff[3]:.2f} e 2 {eff[4]:.2f} {eff[5]:.2f}"

efficienza(pmt)


    
    
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit
from scipy.special import factorial

# ==========================================
# 1. I TUOI DATI
# ==========================================
# Sostituisci questo array con i tuoi valori misurati
dati = np.array([12, 15, 11, 14, 13, 16, 12, 10, 15, 13, 14, 12, 11, 13, 16])


# ==========================================
# 2. DEFINIZIONE DELLA POISSONIANA PER IL FIT
# ==========================================
def poisson(k, mu, A):
    """Distribuzione di Poisson riscalata per l'ampiezza dell'istogramma.

    k: numero di conteggi (deve essere trattato come continuo per il fit)
    mu: valore atteso (parametro del fit) A: fattore di scala per l'ampiezza
    (parametro del fit)
    """
    return A * (mu**k) * np.exp(-mu) / factorial(k)


# ==========================================
# 3. PREPARAZIONE DELL'ISTOGRAMMA
# ==========================================
# Definiamo i bin discreti per conteggi interi
min_val = np.min(dati) - 1
max_val = np.max(dati) + 2
bins = np.arange(min_val, max_val) - 0.5

# Calcoliamo l'istogramma (conteggi e centri dei bin)
counts, bin_edges = np.histogram(dati, bins=bins)
bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2

# Errore Poissoniano sulle frequenze osservate (sqrt(N))
# Impostiamo a 1 se il conteggio è 0 per evitare divisioni per zero
yerr = np.sqrt(counts)
yerr[yerr == 0] = 1

# ==========================================
# 4. FIT CON CURVE_FIT
# ==========================================
# Stima iniziale dei parametri [mu, A]
p0 = [np.mean(dati), len(dati)]

# Fit dei dati (usiamo solo i bin con conteggi per stabilizzare il fit)
mask = counts > 0
popt, pcov = curve_fit(
    poisson,
    bin_centers[mask],
    counts[mask],
    p0=p0,
    sigma=yerr[mask],
    absolute_sigma=True,
)

mu_fit, A_fit = popt
sigma_mu = np.sqrt(pcov[0, 0])  # Errore standard su mu

print(f"Risultato del Fit:")
print(f"mu (conteggio medio) = {mu_fit:.3f} ± {sigma_mu:.3f}")

# ==========================================
# 5. GRAFICO
# ==========================================
plt.figure(figsize=(8, 5))

# Istogramma con barre d'errore
plt.errorbar(
    bin_centers,
    counts,
    yerr=yerr,
    fmt="o",
    color="black",
    ecolor="gray",
    capsize=3,
    label="Dati sperimentali",
)
plt.hist(
    dati,
    bins=bins,
    density=False,
    alpha=0.3,
    color="skyblue",
    edgecolor="black",
)

# Curva di fit
x_fine = np.linspace(min_val, max_val, 200)
plt.plot(
    x_fine,
    poisson(x_fine, mu_fit, A_fit),
    "r-",
    lw=2,
    label=f"Fit Poissoniano\n$\\mu = {mu_fit:.2f} \\pm {sigma_mu:.2f}$",
)

plt.xlabel("Numero di conteggi")
plt.ylabel("Frequenza (N. misurazioni)")
plt.title("Fit Poissoniano dei Conteggi del Rivelatore")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.show()
```
